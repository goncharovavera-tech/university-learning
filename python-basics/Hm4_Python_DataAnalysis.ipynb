## Описание данных
### Таблица ```Hm4_transactions.csv```
##### Описание
Таблица содержит историю транзакций клиентов банка за один год и три месяца.

##### Формат данных

```
customer_id,tr_datetime,mcc_code,tr_type,amount,term_id
111111,15 01:40:52,1111,1000,-5224,111111
111112,15 15:18:32,3333,2000,-100,11122233
...
```
##### Описание полей

 - ```customer_id``` — идентификатор клиента;

 - ```term_id``` — идентификатор терминала;

##### Пояснения:
(\*) Для вычисления максимальной разницы между значениями списка можно использовать функцию np.ptp  
(\**) Если в результате получились значения [1,3,5], то макимальная разница между ними 4 == 5-1.

````python
### Type your code here
med1 = df_transaction["amount"].median()
med2 = df_transaction.dropna()["amount"].median()
df_tran_sort = df_transaction.sort_values("amount")
dubl = ["mcc_code", "tr_type"]
keep = 'last'
df_tran_cleen = df_transaction.sort_values("amount").drop_duplicates(subset=["mcc_code", "tr_type"], keep = "last")
med3 = df_tran_cleen["amount"].median()
print(f"Med1 ={med1}")
print(f"Med2 ={med2}")
print(f"Med3 ={med3}")

medians = [med1, med2, med3]
print(f"Максимальная разница: {round(np.ptp(medians), 2)}")
````
